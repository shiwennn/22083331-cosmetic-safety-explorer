```python
# NOTEBOOK 01: DATA EXPLORATION

import pandas as pd
import os

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 100)

RAW_PATH = '../data/raw'

# ----------------------------------------------------------
# STEP 1: LOAD PRODUCT INFO DATASET
# ----------------------------------------------------------

print("=" * 60)
print("LOADING PRODUCT INFO DATASET")
print("=" * 60)

products = pd.read_csv(os.path.join(RAW_PATH, 'product_info.csv'))

print(f"\nShape: {products.shape[0]} rows × {products.shape[1]} columns")
print(f"\nColumn Names:")
for col in products.columns:
    print(f"  - {col} ({products[col].dtype})")
```

```python
# ----------------------------------------------------------
# STEP 2: PREVIEW PRODUCT DATA
# ----------------------------------------------------------

print("\nFirst 3 rows of product_info:")
display(products.head(3))

print("\nMissing Values in product_info:")
missing = products.isnull().sum()
missing_pct = (missing / len(products) * 100).round(2)

missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
}).query('`Missing Count` > 0').sort_values('Missing %', ascending=False)

if missing_df.empty:
    print("  No missing values found.")
else:
    print(missing_df.to_string())

print("\nDuplicate Product IDs:")
print(products['product_id'].duplicated().sum())

print("\nProduct Categories:")
print(products['primary_category'].value_counts())
```

```python
# ----------------------------------------------------------
# STEP 3: LOAD REVIEW DATASETS
# ----------------------------------------------------------

print("=" * 60)
print("LOADING REVIEW DATASETS")
print("=" * 60)

review_files = [
    'reviews_0-250.csv',
    'reviews_250-500.csv',
    'reviews_500-750.csv',
    'reviews_750-1250.csv',
    'reviews_1250-end.csv'
]

review_dfs = []

for f in review_files:
    path = os.path.join(RAW_PATH, f)
    df = pd.read_csv(path, dtype={'author_id': str}, low_memory=False)
    print(f"  {f}: {df.shape[0]:,} rows × {df.shape[1]} columns")
    review_dfs.append(df)

reviews = pd.concat(review_dfs, ignore_index=True)

print(f"\nCombined Reviews Dataset:")
print(f"  Total rows : {reviews.shape[0]:,}")
print(f"  Total cols : {reviews.shape[1]}")

print(f"\nColumn Names:")
for col in reviews.columns:
    print(f"  - {col} ({reviews[col].dtype})")
```

```python

# ----------------------------------------------------------
# STEP 4: PREVIEW REVIEW DATA
# ----------------------------------------------------------

print("\nFirst 3 rows of reviews:")
display(reviews.head(3))

print("\nMissing Values in reviews:")
missing_r = reviews.isnull().sum()
missing_pct_r = (missing_r / len(reviews) * 100).round(2)

missing_df_r = pd.DataFrame({
    'Missing Count': missing_r,
    'Missing %': missing_pct_r
}).query('`Missing Count` > 0').sort_values('Missing %', ascending=False)

if missing_df_r.empty:
    print("  No missing values found.")
else:
    print(missing_df_r.to_string())

print("\nDuplicate Review Rows:")
print(reviews.duplicated().sum())

print("\nUnique Products in Reviews:")
print(reviews['product_id'].nunique())

print("\nReviews Matched to Product Info:")
matched_reviews = reviews['product_id'].isin(products['product_id']).sum()
print(f"{matched_reviews:,} / {len(reviews):,} reviews")

print("\n" + "=" * 60)
print("EXPLORATION COMPLETE")
print(f"Products dataset : {products.shape[0]:,} products")
print(f"Reviews dataset  : {reviews.shape[0]:,} reviews")
print("=" * 60)
```
